```python
categorias_restaurantes = {
    "Carnes": [
        "steak", "bbq", "hotdogs", "hotdog", "chicken_wings", 
        "butcher", "brazilian", "burgers", "chickenshop", "argentine"
    ],
    "Mariscos": [
        "seafood", "cajun"
    ],
    "Pescado": [
        "sushi", "poke"
    ],
    "Vegetariano": [
        "vegetarian", "salad"
    ],
    "Vegano": [
        "vegan", "raw_food", "juicebars", "acaibowls"
    ],
    "Otro": [
        "cuban", "bars", "venues", "asianfusion", "cocktailbars", 
        "mediterranean", "italian", "spanish", "coffee", "sandwiches", 
        "korean", "hotpot", "wine_bars", "pizza", "tradamerican", 
        "sportsbars", "newamerican", "lounges", "mexican", "indpak", 
        "greek", "lebanese", "foodstands", "jazzandblues", "french", 
        "japanese", "peruvian", "bubbletea", "ramen", "tacos", 
        "mideastern", "arabian", "portuguese", "puertorican", "tapas", 
        "breakfast_brunch", "desserts", "cafes", "foodtrucks", "noodles", 
        "izakaya", "empanadas", "thai", "latin", "chinese", "caribbean", 
        "musicvenues", "modern_european", "turkish", "basque", 
        "fooddeliveryservices", "beer_and_wine", "karaoke", "irish", 
        "irish_pubs", "gastropubs", "soulfood", "tapasmallplates", 
        "persian", "hotels", "restaurants", "beerbar", "polish", 
        "dimsum", "panasian", "nicaraguan", "taiwanese", "buffets", 
        "bakeries", "comfortfood", "food_court"
    ],
    None: [] # Representa el valor 'nan' de tu lista
}
```

```python
mejores_opciones = []

for categoria, alias_list in categorias_restaurantes.items():
    categoria_filtro = '|'.join(alias_list)
    
    #aplicar filtro
    df_filtrado = df_yelp[df_yelp['alias'].str.contains(categoria_filtro, case=False, na=False)]

    #filtro top 1
    df_top = df_filtrado.sort_values(by='rating', ascending=False).iloc[0]
    
    # consolidado
    mejores_opciones.append({
        'preferencias_alimenticias': categoria,
        'nombre_restaurante': df_top['title'],
        'rating': df_top['rating'],
        'review_count': df_top['review_count'],
        'direccion': df_top['location_address1'],
    })
```

```python
df_resturantes_recomendados = pd.DataFrame(mejores_opciones)
```
